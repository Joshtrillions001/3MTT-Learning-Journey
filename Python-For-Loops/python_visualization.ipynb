```python
def sales_amount(transactions):
    counter = 0
    for item in transactions:
        
      if item > 100:
         counter += 1
    return counter

sales_amount((1,40,101,100,300,200,150))
```

```python
def sales_amount(transactions):
    counter = 0
    for item in transactions:
        
      if item > 100:
         counter += 1
    return counter

sales_amount((50, 150, 200, 10))
```

```python
def purchases_100(transactions):
    return sum(1 for x in transactions if x > 100)

purchases_100((50, 150, 200, 10))
```
